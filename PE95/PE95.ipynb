```python
from array import array

def chains(N):
    a = array('i', [1]) * N
    for f in range(2, N):
        for i in range(f*2, N, f):
            a[i] += f

    for x in range(N):
        seen = []
        while x < N and a[x] != -1:
            seen.append(x)
            a[x], x = -1, a[x]
        if x in seen:
            yield seen[seen.index(x):]

print(min(max(chains(10**6), key=len)))
```
